# LCAO und Tight-Binding in endlichen Gittern

Wir verwenden pro Atom ein normiertes $s$-Orbital und vernachlässigen die Überlappung der Basisorbitale, also $\langle m|n\rangle=\delta_{mn}$. Die Wellenfunktion ist $\psi=\sum_i c_i\phi_i$. Die Gitter haben offene Ränder (keine periodischen Ringe oder Torus). Es werden nur nächste Nachbarn gekoppelt; das Hopping-Matrixelement ist $-V_{ss}$ und das onsite-Matrixelement $\epsilon_0$.

Die interaktiven Modelle zeigen die Basis, Hamiltonmatrix, Eigenzustände, die analytischen Energien und eine numerische Probe von $Hc=Ec$.

## 1D: Lineare Kette

Für eine Kette aus $N$ Atomen ist $i=n$ der Basisindex und $\phi_n(x)=\phi_s(x-na)$. Die Eigenzustände sind stehende Sinuswellen

$$c_n^{(l)}=\sqrt{\frac{2}{N+1}}\sin\left(\frac{nl\pi}{N+1}\right),\qquad E_l=\epsilon_0-2V_{ss}\cos\left(\frac{l\pi}{N+1}\right).$$

Die $N$ Zustände liegen auf dem 1D-Band $E(k)=\epsilon_0-2V_{ss}\cos(ka)$ zwischen $k=0$ und $\pi/a$. Seine Breite beträgt $4V_{ss}$. Im Eigenvektor ist die Kreisgröße proportional zu $|c_n|$; blau steht für positive und rot für negative Koeffizienten.

## 2D: Quadratgitter

Die Atome werden zeilenweise nummeriert: $i=(n_y-1)N_x+n_x$. Für jeden Zustand gibt es zwei Quantenzahlen $(l_x,l_y)$; der Eigenvektor ist das Produkt zweier 1D-Sinuswellen:

$$c_{n_x,n_y}=\frac{2}{\sqrt{(N_x+1)(N_y+1)}}\sin\left(\frac{n_xl_x\pi}{N_x+1}\right)\sin\left(\frac{n_yl_y\pi}{N_y+1}\right).$$

$$E_{l_x,l_y}=\epsilon_0-2V_{ss}\left[\cos\left(\frac{l_x\pi}{N_x+1}\right)+\cos\left(\frac{l_y\pi}{N_y+1}\right)\right].$$

Vertikale Nachbarn unterscheiden sich im Index um $N_x$. Die Dispersionsfläche $E(k_x,k_y)=\epsilon_0-2V_{ss}(\cos k_xa+\cos k_ya)$ hat eine Bandbreite von $8V_{ss}$. Im Eigenvektor ist die Kreisgröße proportional zu $|c_i|$; blau steht für positive und rot für negative Koeffizienten. Entartete Energieniveaus stehen nebeneinander.
